```python
import datetime

class User:
    def __init__(self, name):
        self.name = name
        self.created_at = datetime.datetime.now()

    def show_created_time(self):
        return self.created_at.time()

u1 = User("Ali")
u2 = User("Vali")

print(u1.show_created_time())
print(u2.show_created_time())
```

```python
import datetime

birth_input = input("Enter birth date (YYYY-MM-DD): ")

convert_to_date = datetime.date.fromisoformat(birth_input)
today = datetime.date.today()

difference = today - convert_to_date
total_days = difference.days

years = total_days // 365
days = total_days % 365
month = days // 30

print(f"Men {years} yil yashaganman")
print(f"Men {days} kun yashaganman")
print(f"Men {month} oy yashaganman")
```

```python
import datetime
birth_input = input("Enter birth day (MM-DD): ")

today = datetime.date.today()
parts = birth_input.split("-")
month = int(parts[0]) #user birth month
day = int(parts[1]) # user birth day

year = today.year  #  shu year

birthday_this_year = datetime.date(year, month, day) # users lest birth day

if birthday_this_year >= today:
    next_birthday = birthday_this_year
else:
    next_birthday = datetime.date(year + 1, month, day)

days_left = next_birthday - today

print (days_left)
```
