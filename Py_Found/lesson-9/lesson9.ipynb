```python
class BankAccount:
    def __init__(self):
        self.balance = 0

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            print("Not enough balance")
        else:
            self.balance -= amount

    def show_balance(self):
        print(self.balance)

acc = BankAccount()
acc.deposit(100)
acc.withdraw(30)
acc.show_balance()

```

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        self.grades = []

    def study(self):
        print(f"{self.name} is studying")

    def add_grade(self, grade):
        self.grades.append(grade)

    def get_average(self):
        if len(self.grades) == 0:
            return 0
        return sum(self.grades) / len(self.grades)

    def info(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Grades:", self.grades)
        print("Average:", self.get_average())
```

```python
student1 = Student("Mavlono", 22)
student1.study()
student1.add_grade(80)
student1.add_grade(90)
student1.add_grade(70)
student1.info()

```

```python
student2 = Student("Ali", 20)
student2.add_grade(100)
student2.info()
```

```python
class Car:
    def __init__ (self,  year, speed, brand):
        self.year = year
        self.speed = speed
        self.brand = brand
    def drive (self):
        print (f"{self.brand} is driving")
    def accelerate(self, amount):
        self.speed += amount
    def brake(self, amount):
        self.speed -= amount
    def show_status(self):
        print("Brand", self.brand)
        print("year", self.year)
        print("speed", self.speed)

Car1 = Car(2024,142,"BMW")
Car1.accelerate(50)
Car1.brake(30)
Car1.drive()
Car1.show_status()
Car2 = Car(2020, 160, "Audi")
Car3 = Car(2022,170,"Tesla")

```

```python
class Student:
    def __init__(self, name,score):
        self.name = name
        self.score = score
    def grade (self):
        if self.score == 100:
            return "A+"
        elif self.score > 90:
            return "A"
        elif self.score > 70:
            return "B"
        elif self.score > 50:
            return "C"
        else: ## condition berilmiydi
            return "Failed"

    def show_result(self):
        print(f"Student name is {self.name} and his score is {self.score}. Finally hi is {self.grade()} grade")

test1 = Student("Mavlonbek",45)
test2 = Student("Jamshid", 75)

test2.show_result() ## Mavlonbek
```
