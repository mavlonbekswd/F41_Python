| Method    | Purpose                 |
| --------- | ----------------------- |
| append()  | add element to end      |
| insert()  | add element at position |
| extend()  | add multiple elements   |
| remove()  | remove value            |
| pop()     | remove by index         |
| clear()   | remove all elements     |
| index()   | find position            |
| count()   | count occurrences       |
| sort()    | sort list               |
| reverse() | reverse list            |
| copy()    | copy list               |
