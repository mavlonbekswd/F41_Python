np.concatenate()
np.vstack()
np.hstack()
np.column_stack()

```python
import numpy as np

a = np.array([1,2,3])
b = np.array([4,5,6])

print(np.concatenate([a,b]))
```

```python
#2D Concatenate
a = np.array([
    [1,2],
    [3,4]
])

b = np.array([
    [5,6]
])

print(np.concatenate([a,b], axis=0)) #vertical rows qo‘shadi.
#a.shape -> (2,2)
#b.shape -> (1,2)
#Columns soni bir xil bo‘lishi kerak
```

```python
a = np.array([
    [1,2],
    [3,4]
])

b = np.array([
    [5],
    [6]
])

np.concatenate([a,b], axis=1) #garizontal col qoshadi
```

```python
#vstack() ← vertical stack

a = np.array([1,2,3])
b = np.array([4,5,6])

np.vstack([a,b])

#1D array’larni avtomatik 2D rowga aylantiradi.
```

```python
#hstack() ← horizontal stack

a = np.array([1,2,3])
b = np.array([4,5,6])

np.hstack([a,b])
```

```python
#2D
a = np.array([
 [1],
 [2]
])

b = np.array([
 [3],
 [4]
])

np.hstack([a,b])
```

```python
name_id = np.array([101,102,103])
score = np.array([80,90,70])

np.column_stack([name_id, score])
```

```python
a = np.array([
 [1,2],
 [3,4]
])

b = np.array([
 [5,6]
])

print(np.vstack([a,b]))
```
