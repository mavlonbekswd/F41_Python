```python
import numpy as np

a = np.array([1,2,3])
b = np.array([4,5,6])

print(np.concatenate([a,b]))
```

```python
#2D Concatenate
a = np.array([
    [1,2],
    [3,4]
])

b = np.array([
    [5,6]
])

print(np.concatenate([a,b], axis=0)) #vertical rows qo‘shadi.
#a.shape -> (2,2)
#b.shape -> (1,2)
#Columns soni bir xil bo‘lishi kerak
```

```python
a = np.array([
    [1,2],
    [3,4]
])

b = np.array([
    [5],
    [6]
])

np.concatenate([a,b], axis=1) #garizontal col qoshadi
```
