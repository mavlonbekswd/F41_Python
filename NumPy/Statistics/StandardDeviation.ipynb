🔥 Standard Deviation (STD)

Agar Mean: "O‘rtacha qancha?" degan savolga javob bersa,

STD:
"Qiymatlar Mean'dan qanchalik uzoqqa tarqalgan?" degan savolga javob beradi.

```python
import numpy as np

a = np.array([98,100,102,101,99])
b = np.array([20,180,50,170,80])

print(np.std(a))
print(np.std(b))
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

sales = np.random.normal(
    loc=1000,
    scale=200,
    size=1000
)

print("Mean:", np.mean(sales))
print("STD :", np.std(sales))

plt.hist(sales, bins=30)
plt.axvline(np.mean(sales), linestyle="--")
plt.show()
```

```python
import numpy as np

sales1 = np.array([98,100,102,101,99])
sales2 = np.array([20,180,50,170,80]) #orasidagi faqr outputga chiqadi

print(np.mean(sales1))
print(np.std(sales1))

print(np.mean(sales2))
print(np.std(sales2))

#Qiymatlar mean'dan o'rtacha 64.18 birlik uzoqlikda.
#data chudayam tarqoq bo'lsa orasidangi sonlar kattalashi
```
