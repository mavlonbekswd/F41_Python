```python
import numpy as np
import matplotlib.pyplot as plt

orders = np.array([1, 2, 3, 4, 10])

mean_value = np.mean(orders)
median_value = np.median(orders)

plt.figure(figsize=(8, 4))

plt.scatter(orders, np.zeros_like(orders), s=100)

plt.axvline(mean_value, linestyle="--", label=f"Mean = {mean_value}")
plt.axvline(median_value, linestyle="--", label=f"Median = {median_value}")

plt.title("Mean vs Median with Outlier")
plt.xlabel("Order Value")
plt.yticks()
plt.legend()
plt.show()
```

```python
orders_without_outlier = orders[orders < 50]

print("Original mean:", np.mean(orders))
print("Original median:", np.median(orders))

print("Without outlier mean:", np.mean(orders_without_outlier))
print("Without outlier median:", np.median(orders_without_outlier))
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

sales = np.random.normal(
    loc=1000,
    scale=200,
    size=1000
)

mean_value = np.mean(sales)
median_value = np.median(sales)
sales = np.append(
    sales,
    [10000]
)

plt.figure(figsize=(15,5))

plt.hist(
    sales,
    bins=100
)

plt.axvline(
    mean_value,
    linestyle="--",
    label=f"Mean = {mean_value:.2f}"
)

plt.axvline(
    median_value,
    linestyle="--",
    label=f"Median = {median_value:.2f}"
)

plt.title("Sales Distribution")
plt.xlabel("Sales (£)")
plt.ylabel("Frequency")

plt.legend()
plt.show()
```
