Array elementlarini bittalab aylanish (loop qilish).

```python
import numpy as np

arr = np.array([10,20,30])

for x in arr:
    print(x)
```

```python
#2D Array Iteration, dimension bo‘yicha yuradi.
arr = np.array([
    [1,2,3],
    [4,5,6]
])

for x in arr:
    print(x)
```

```python
#2D ichidagi elementlarni aylanish Nested loop kerak.

arr = np.array([
    [1,2,3],
    [4,5,6]
])

for row in arr:
    for item in row:
        print(item)
```

```python
#np.nditer() proficonal iteration
###barcha elementlarni bittalab yuradi.
arr = np.array([
    [1,2],
    [3,4]
])

for x in np.nditer(arr):
    print(x)

```
