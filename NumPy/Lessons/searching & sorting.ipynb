#SEARCH

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 4, 4])

x = np.where(arr == 4)

print(x)

#(array([3, 5, 6]),) yani bular 4 raqami bor bo'lgan indexlar
```

```python
arr = np.array([50, 10, 40, 20])

np.argsort(arr) #Chunki sorted orderdagi indexlarni qaytaradi.

arr[1] = 10
arr[3] = 20
arr[2] = 40
arr[0] = 50
```

```python
arr = np.array([70, 10, 50, 20])

print(np.argsort(arr))
```

```python
sales = np.array([1200, 800, 5000, 300, 7000])

print(np.max(sales))
print(np.where(sales == 800))
```

```python
import numpy as np

arr = np.array([10, 14, 93, 41, 8, 7])

x = arr[np.where(arr%2 == 1)] #odd numbers
x = np.where(arr%2 == 0) #even numbers

print(x)
```

```python

#“Array sorted bo‘lib turganda yangi qiymat qayerga qo‘yilsa tartib buzilmaydi?”

#Juda muhim shart
#Array SORTED bo‘lishi kerak.
#Agar: [40,10,30] bersang — natija unreliable bo‘ladi.
arr = np.array([1, 3, 5, 7, 40,56,72])

print(np.searchsorted(arr, 50))
```

```python
import numpy as np

arr = np.array([1, 3, 5, 7])

x = arr[np.searchsorted(arr, [2, 4, 6])]

print(x)
```

```python
#side="left" va side="right" aslida duplicate qiymatlar bo‘lganda farq qiladi.
arr = np.array([1, 2, 2, 2, 3])

print(np.searchsorted(arr, 2, side= "right"))
```

```python
arr = np.array([10,  20, 30, 40])

print(np.searchsorted(arr, [15, 35]))
```
