```python
import numpy as np

print(np.random.randint(1, 10))
```

```python
#Bir nechta random son
arr = np.random.randint(1, 100, size=5)

print(arr)
```

```python
#2D random array
arr = np.random.randint(1, 10, size=(3,3))

print(arr)
```

```python
#rand() faqat butun sonlarni oladi
arr = np.random.rand(5)

#2D random matrix
np.random.rand(2,3)

print(arr)
```

```python
#np.random.choice() 🔥

#Bu juda ko‘p ishlatiladi.

#Array ichidan random tanlaydi.

colors = ["Red", "Blue", "Green"]

print(np.random.choice(colors))
```

```python
#Bir nechta tanlash
colors = ["Red", "Blue", "Green"]

print(np.random.choice(colors, size=5))
```

```python
from numpy import random

#probability'lar yig'indisi 1 bo'lishi kerak.

x = random.choice([3, 5, 7, 9], p=[0.3, 0.1, 0.6, 0.0], size=(100))

print(x)
```
