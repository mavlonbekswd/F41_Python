```python
#Shuffle means changing arrangement of elements in-place. i.e. in the array itself.

from numpy import random
import numpy as np

arr = np.array([1, 2, 3, 4, 5])

#shuffle() original array'ni o‘zgartiradi.

random.shuffle(arr)

print(arr)
```

```python
#permutation() original arrayni o‘zgartirmaydi.

from numpy import random
import numpy as np

arr = np.array([1,2,3,4,5])

new_arr = random.permutation(arr)

print(new_arr)
print(arr)
```
