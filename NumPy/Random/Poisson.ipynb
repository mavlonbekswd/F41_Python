```python
import numpy as np

np.random.seed(42)

days = 30

# 1) Poisson Distribution
# Har kuni nechta call keladi?
# O'rtacha kuniga 120 ta call
calls_per_day = np.random.poisson(lam=120, size=days)

# 2) Binomial Distribution
# Kelgan calllardan nechtasi successful resolved bo'ladi?
# Har bir call 75% ehtimol bilan hal bo'ladi
resolved_calls = np.random.binomial(n=calls_per_day, p=0.75)

# 3) Normal Distribution
# Har bir kun uchun average call duration
# O'rtacha 6 minut, std = 1.5 minut
avg_call_duration = np.random.normal(loc=6, scale=1.5, size=days)

# Negative duration chiqib qolmasligi uchun
avg_call_duration = np.round(np.clip(avg_call_duration, 1, None), 2)

#"Clip nima qiladi?
#1 dan kichik bo'lsa → 1 qil
#1 dan katta bo'lsa → tegma"

# Extra calculations
unresolved_calls = calls_per_day - resolved_calls
total_minutes = calls_per_day * avg_call_duration

print("Calls per day:")
print(calls_per_day)

print("\nResolved calls:")
print(resolved_calls)

print("\nUnresolved calls:")
print(unresolved_calls)

print("\nAverage call duration:")
print(avg_call_duration)

print("\nTotal working minutes per day:")
print(np.round(total_minutes, 2))

print("\n--- Summary ---")
print("Total calls:", calls_per_day.sum())
print("Total resolved:", resolved_calls.sum())
print("Resolution rate:", round(resolved_calls.sum() / calls_per_day.sum() * 100,2), "%")
print("Average calls per day:", round(calls_per_day.mean(), 2))
print("Average duration:", round(avg_call_duration.mean(), 2), "minutes")
```
