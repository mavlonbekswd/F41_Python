```python
from numpy import random
#defoult xolatda faqat 0 va 1 orasida bo'ladi
x = random.uniform()

print(x)
```

```python
#O'zimiz interval beramiz
x = random.uniform(10, 20)

print(x)
```

```python
#Ko‘p son yaratish
x = random.uniform(10, 20, size=5)

print(x)
```

```python
#2D Array
x = random.uniform(0, 100, size=(2,3))

print(x)
```

```python
#Visualization of Uniform Distribution

from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

sns.displot(random.uniform(size=1000), kind="kde")

plt.show()
```
