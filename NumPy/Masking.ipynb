```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print(arr > 25)
```

```python
10 > 25 → False
20 > 25 → False
30 > 25 → True
40 > 25 → True
50 > 25 → True
```

```python
print(arr[arr>25])
```

```python
arr = np.array([
    [10,20,30],
    [40,50,60]
])

print(arr[arr > 30])
```

```python
arr = np.array([10, 15, 20, 25, 30, 35, 40])


newarray = (arr > 20) & (arr<40)

justnumber = (arr[newarray])
oddarray = (justnumber % 2 != 0)
finalarr =(justnumber[oddarray])

print(finalarr)
```

```python
import numpy as np

arr = np.array([10, 15, 20, 25, 30, 35, 40])

result = arr[
    (arr > 20) &
    (arr < 40) &
    (arr % 2 != 0)
]

print(result)
```
